```python
from langchain_anthropic import ChatAnthropic
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatAnthropic(model="claude-haiku-4-5-20251001", max_tokens=20, temperature=0)
```

```python
result = model.invoke("Write a code to print 'Hello World' in Python")
print(result.content)
```
