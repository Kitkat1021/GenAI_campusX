```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatOpenAI(model="gpt-4",
                   temperature=2, # Temperature controls the creativity
                   max_tokens=10 # limit the tokens
)
```

```python
result = model.invoke("What is the capital of Ireland?")
```

```python
print(result)         # This will print the entire response object, which may include metadata
print(result.content) # Need only the response content
```

```python
result_1 = model.invoke("Write a poem about sky in 4 lines")
print(result_1.content) # when temperture is 0
```

```python
result_2 = model.invoke("Write a poem about sky in 4 lines")
print(result_2.content) # Temperature is set to 2 for this case
```
